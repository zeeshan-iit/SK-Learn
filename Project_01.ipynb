```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, accuracy_score
```

```python
df = pd.read_csv(r"E:\SK-Learn\Student.csv")
df.head()
```

```python
X = df[['Quiz01 [10]', 'Assignment01 [8]', 'Midterm Exam [20]',
       'Assignment02 [12]', 'Assignment03 [25]', 'Final Exam [35]']]
Y = df['Total [100]']
```

```python
model = LinearRegression()
model.fit(X,Y)
predicted_marks = model.predict(X)
```

```python
# valid regression matrics.
MAE = mean_absolute_error(Y, predicted_marks)
MSE = mean_squared_error(Y, predicted_marks)
RMSE = np.sqrt(MSE)
R2_Score = r2_score(Y, predicted_marks)

print("Mean Absolute Errror : ", round(MAE, 2))
print("Mean Square Errror : ", round(MSE, 2))
print("Root Mean Square Errror : ", round(RMSE, 2))
print("R2 Score : ", round(R2_Score, 2))
```

```python
# Histogram
plt.hist(
    predicted_marks,
    edgecolor = 'black',
    bins = 20,
    color = 'aqua'
)
plt.title("Predicted Marks by Model")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
# Scatter plot
plt.scatter(X["Final Exam [35]"], Y)
plt.show()
```
