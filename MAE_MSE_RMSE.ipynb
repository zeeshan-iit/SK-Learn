```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
```

```python
x = [70, 60, 90, 50, 100, 80]
y = [70, 50, 85, 48, 100, 80]
```

```python
mae = mean_absolute_error(x, y)
mse = mean_squared_error(x, y)
rmse = np.sqrt(mse)
```

```python
print(mae, mse, rmse)
```

```python
# MAE:	Mean Absolute Error	Average distance from the truth.
# MSE:	Mean Squared Error	Penalizes big mistakes more.
# RMSE:	Root Mean Squared Error	Error in the original units (like "dollars" or "meters").
```
