```python
from sklearn.metrics import confusion_matrix
```

```python
x = [1, 0, 0, 1, 0, 1, 1, 1, 0, 1, 1]
y = [1, 1, 0, 0, 1, 1, 1, 0, 1, 1, 0]
```

```python
cm = confusion_matrix(x,y)
```

```python
print(cm)
```

```python
# The matrix is a 2x2 grid:
# True Positive (TP): The model said "Dog," and it was a Dog. (Correct)
# True Negative (TN): The model said "Cat," and it was a Cat. (Correct)
# False Positive (FP): The model said "Dog," but it was a Cat. (False Alarm)
# False Negative (FN): The model said "Cat," but it was a Dog. (Missed Target)
```

```python
#                |  Predicted: No	 |       Predicted: Yes
# Actual: No	|   1 (Correct)	     |      3 (Mistake)
# Actual: Yes   |   3 (Mistake)  	 |     4 (Correct)
```
