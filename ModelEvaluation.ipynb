```python
from sklearn.metrics import precision_score, accuracy_score, f1_score, recall_score
```

```python
y_true = [1, 0, 1, 1, 1, 0, 0, 1, 0]
y_predict = [1, 0, 0, 1, 0, 1, 0, 0, 1]
```

```python
print("Accuracy Score : ", accuracy_score(y_true, y_predict))
print("Precision Score : ", precision_score(y_true, y_predict))
print("F1 Score : ", f1_score(y_true, y_predict))
print("Recall Score : ", recall_score(y_true, y_predict))
```

#Accuracy: 	How often is the model right overall?	Maximize overall hits.
#Precision:	How sure are we when the model says "Yes"?	Minimize False Alarms.
#Recall:	Did we miss any "Yes" cases?	Catch every target.
#F1-Score:	Is there a good balance of both?	Find a middle ground
