```python
from imblearn.over_sampling import SMOTE
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv("D:\AI\code\seattle-weather.csv")
df.columns
```

```python
X = df[['precipitation', 'temp_max', 'temp_min', 'wind']]
Y = df['weather']
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    X, Y, test_size = 0.2, random_state = 42
)
```

```python
classification = y_train.value_counts()
print(f"Data set imbalancing before smote filter{classification}")
```

```python
smote = SMOTE(random_state = 42)
x_resample, y_resample = smote.fit_resample(x_train, y_train)
```

```python
print(f"The dataset after appling the smote filter : {y_resample.value_counts()}")
```
