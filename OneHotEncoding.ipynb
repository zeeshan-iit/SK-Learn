```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
import numpy as np
```

```python
data = pd.DataFrame({
    "Gender" : ["Male", "Female", "Female","Male", "Female", "Male","Male", "Female", "Female"],
    "Pass" : ["Yes", "No", "Yes", "No", "No", "No", "Yes", "Yes", "No"]
})
```

```python
encoding = OneHotEncoder(sparse_output = False)
```

```python
data[['Gender_Female','Gender_Male','Pass_NO','Pass_Yes']] = encoding.fit_transform(data[['Gender','Pass']])
```
